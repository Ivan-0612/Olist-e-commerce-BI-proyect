# Transformar en modelo estrella

Actualmente las tablas tienen formato copo de nieve, para cumplir con la metodología medallion voy a transformarlo en un modelo de estrella, con una tabla de hechos y varias de dimensiones.

```python
# Dimensiones -> Customers, Sellers, Products, Date

# 1. Customers
geo_clean = silver_geolocation.groupBy("geolocation_zip_code_prefix").agg(
    F.first("geolocation_city").alias("geo_city"),
    F.first("geolocation_state").alias("geo_state"),
    F.avg("geolocation_lat").alias("latitude"),
    F.avg("geolocation_lng").alias("longitude")
) #para evitar duplicar claves y dejar un unico dato por codigo postal

gold_dim_customers = silver_customers.join(
    geo_clean,
    silver_customers["customer_zip_code_prefix"] == geo_clean["geolocation_zip_code_prefix"],
    how="left"
).select(
    F.col("customer_id"),
    F.col("customer_unique_id"),
    F.col("customer_zip_code_prefix").alias("zip_code"),
    F.coalesce(F.col("customer_city"), 
    F.col("geo_city")).alias("city"),
    F.coalesce(F.col("customer_state"), 
    F.col("geo_state")).alias("state"),
    F.col("latitude"),
    F.col("longitude")
)


# 2. Sellers
gold_dim_sellers = silver_sellers.join(
    geo_clean,
    silver_sellers["seller_zip_code_prefix"] == geo_clean["geolocation_zip_code_prefix"],
    how="left"
).select(
    F.col("seller_id"),
    F.col("seller_zip_code_prefix").alias("zip_code"),
    F.coalesce(F.col("seller_city"), F.col("geo_city")).alias("city"),
    F.coalesce(F.col("seller_state"), F.col("geo_state")).alias("state"),
    F.col("latitude"),
    F.col("longitude")
)

# 3. Products
gold_dim_products = silver_products.select(
    F.col("product_id"),
    F.col("product_category_name").alias("category"),
    F.col("product_name_lenght").alias("name_length"),
    F.col("product_description_lenght").alias("description_length"),
    F.col("product_photos_qty").alias("photos_count"),
    F.col("product_weight_g").alias("weight_grams"),
    F.col("product_length_cm").alias("length_cm"),
    F.col("product_height_cm").alias("height_cm"),
    F.col("product_width_cm").alias("width_cm")
)

# 4. Date -> creada de 0 para el Time Intelligence de Power BI
import pyspark.sql.functions as F

start_date = "2016-01-01"
end_date = "2018-12-31"

gold_dim_date = spark.sql(f"""
    SELECT explode(sequence(to_date('{start_date}'), to_date('{end_date}'), interval 1 day)) AS date_key
""").select(
    F.col("date_key"),
    F.year("date_key").alias("year"),
    F.quarter("date_key").alias("quarter"),
    F.month("date_key").alias("month"),
    # Mantenemos las columnas originales para no romper el modelo
    F.date_format("date_key", "MMMM").alias("month_name"),
    F.dayofmonth("date_key").alias("day_of_month"),
    F.dayofweek("date_key").alias("day_of_week"),
    F.date_format("date_key", "E").alias("day_name"),
    
    # Añadimos los nuevos campos para ordenación
    F.date_format("date_key", "yyyy-MM").alias("year_month"),
    F.date_format("date_key", "MM - MMMM").alias("month_ordered")
) #una fila por cada día entre el inicio y el fin
```

```python
# Tabla de hechos -> orders + order_items + order_payments + order_reviews

# 1. Consolidar pagos por order_id (sumar importe total y extraer el método de pago principal)
payments_summary = silver_order_payments.groupBy("order_id").agg(
    F.sum("payment_value").alias("total_payment_value"),
    F.first("payment_type").alias("primary_payment_type"),
    F.max("payment_installments").alias("max_installments")
)

# 2. order_review
reviews_summary = silver_order_reviews.select(
    F.col("order_id"),
    F.col("review_score").cast("int").alias("review_score")
)

# 3. Cruzar todo sobre order_items + orders
gold_fact_orders = silver_order_items.join(
    silver_orders, on="order_id", how="inner"
).join(
    payments_summary, on="order_id", how="left"
).join(
    reviews_summary, on="order_id", how="left"
).select(
    # Claves foráneas 
    F.col("order_id"),
    F.col("order_item_id"),
    F.col("customer_id"),
    F.col("product_id"),
    F.col("seller_id"),
    F.col("order_purchase_timestamp").alias("order_date_key"),
    F.col("order_delivered_customer_date").alias("delivered_date_key"),
    
    # Atributos de estado
    F.col("order_status"),
    F.col("primary_payment_type"),
    
    # Métricas numéricas 
    F.col("price").cast("decimal(10,2)").alias("sales_amount"),
    F.col("freight_value").cast("decimal(10,2)").alias("freight_amount"),
    F.col("total_payment_value").cast("decimal(10,2)").alias("order_payment_amount"),
    F.col("max_installments").cast("int").alias("installments"),
    F.datediff(F.col("order_delivered_customer_date"), F.col("order_purchase_timestamp")).alias("delivery_days"),
    F.col("review_score")
)

# Tratamiento outliers (detectado en visualización)
fact_filtered_time = gold_fact_orders.filter( #eliminar pedidos residuales y meses vacíos
    (F.col("order_date_key") >= "2017-01-01") & 
    (F.col("order_date_key") <= "2018-08-31")
)

quantiles = fact_filtered_time.approxQuantile("sales_amount", [0.25, 0.75], 0.01)
q1, q3 = quantiles[0], quantiles[1]
iqr = q3 - q1

lower_limit = max(0.0, q1 - 1.5 * iqr)
upper_limit = q3 + 3.0 * iqr #3 para evitar compras altas legítimas

gold_fact_orders = fact_filtered_time.filter(
    (F.col("sales_amount") >= lower_limit) & 
    (F.col("sales_amount") <= upper_limit)
)
```

```python
# Guardar en tablas delta
spark.conf.set("spark.sql.parquet.vorder.enabled", "true")
spark.conf.set("spark.microsoft.delta.optimizeWrite.enabled", "true")

# Dimensiones
gold_dim_customers.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("gold_dim_customers")
gold_dim_sellers.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("gold_dim_sellers")
gold_dim_products.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("gold_dim_products")
gold_dim_date.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("gold_dim_date")

# Tabla hechos
gold_fact_orders.write.format("delta").mode("overwrite").option("overwriteSchema", "true").saveAsTable("gold_fact_orders")
```
